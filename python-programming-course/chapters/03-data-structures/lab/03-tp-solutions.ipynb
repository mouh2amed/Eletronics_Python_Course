```python
# Shared dataset — same as in the lab notebook
sensor_data = [
    {"timestamp": "2024-01-15 08:00:00", "sensor_id": "S1", "temperature_c": 22.3, "pressure_kpa": 101.2, "status": "OK"},
    {"timestamp": "2024-01-15 08:00:00", "sensor_id": "S2", "temperature_c": 23.1, "pressure_kpa": 101.5, "status": "OK"},
    {"timestamp": "2024-01-15 08:15:00", "sensor_id": "S1", "temperature_c": 22.8, "pressure_kpa": 101.3, "status": "OK"},
    {"timestamp": "2024-01-15 08:15:00", "sensor_id": "S2", "temperature_c": 23.4, "pressure_kpa": 101.4, "status": "OK"},
    {"timestamp": "2024-01-15 08:30:00", "sensor_id": "S1", "temperature_c": 23.2, "pressure_kpa": 101.1, "status": "OK"},
    {"timestamp": "2024-01-15 08:30:00", "sensor_id": "S2", "temperature_c": -999.0, "pressure_kpa": 101.6, "status": "ERROR"},
    {"timestamp": "2024-01-15 08:45:00", "sensor_id": "S1", "temperature_c": 23.5, "pressure_kpa": 101.0, "status": "OK"},
    {"timestamp": "2024-01-15 08:45:00", "sensor_id": "S2", "temperature_c": 24.1, "pressure_kpa": 101.7, "status": "OK"},
    {"timestamp": "2024-01-15 09:00:00", "sensor_id": "S1", "temperature_c": 24.0, "pressure_kpa": 100.9, "status": "OK"},
    {"timestamp": "2024-01-15 09:00:00", "sensor_id": "S2", "temperature_c": 24.3, "pressure_kpa": 101.8, "status": "OK"},
]
print(f"{len(sensor_data)} records loaded.")
```

---

## Solution — Exercise 1: List Basics


```python
# 1. First record
print("First record:", sensor_data[0])

# 2. Last record
print("\nLast record:", sensor_data[-1])

# 3. Slice [2:5] gives indices 2, 3, 4
print("\nRecords 3-5 (index 2-4):")
for row in sensor_data[2:5]:
    print(" ", row)

# 4. All temperatures via list comprehension
all_temperatures = [row["temperature_c"] for row in sensor_data]
print(f"\nAll temperatures: {all_temperatures}")

# 5. Stats
print(f"Count  : {len(all_temperatures)}")
print(f"Min    : {min(all_temperatures):.1f} °C")
print(f"Max    : {max(all_temperatures):.1f} °C")
print(f"Sum    : {sum(all_temperatures):.1f} °C")
```

**Note**: `min(all_temperatures)` returns `-999.0` because the faulty reading is still present. In Exercise 2 we clean it out. This is intentional — students should see that statistics on uncleaned data give meaningless results.


---

## Solution — Exercise 2: Data Cleaning


```python
# 1. Copy
cleaned = all_temperatures.copy()

# 2. Remove fault value
cleaned.remove(-999.0)

# 3. Sort ascending in place
cleaned.sort()

# 4. Print
print(f"Cleaned (ascending): {cleaned}")

# 5. Descending using sorted()
cleaned_desc = sorted(cleaned, reverse=True)

# 6. Verify they are independent
print(f"Cleaned ascending : {cleaned}")
print(f"Cleaned descending: {cleaned_desc}")
print(f"Same object? {cleaned is cleaned_desc}")  # False — they are different lists

# 7. Stats after cleaning
print(f"\nAfter cleaning:")
print(f"  Min    : {min(cleaned):.1f} °C")
print(f"  Max    : {max(cleaned):.1f} °C")
print(f"  Average: {sum(cleaned) / len(cleaned):.2f} °C")
```

**Key point**: `cleaned is cleaned_desc` is `False` — they occupy different locations in memory. Modifying `cleaned` after this point will not affect `cleaned_desc`. This distinction between identity (`is`) and equality (`==`) is important in Python.

## Solution — Exercise 3: List Comprehensions


```python
# 1. Valid records
valid_readings = [
    row for row in sensor_data
    if row["status"] == "OK" and row["temperature_c"] > -100
]
print(f"Valid records: {len(valid_readings)}")

# 2. S1 temperatures from valid readings
s1_temps = [
    row["temperature_c"] for row in valid_readings
    if row["sensor_id"] == "S1"
]
print(f"S1 temps: {s1_temps}")

# 3. S1 temperatures in Fahrenheit
temps_fahrenheit = [t * 9 / 5 + 32 for t in s1_temps]
print(f"S1 temps (°F): {[round(t, 1) for t in temps_fahrenheit]}")

# 4. Timestamps above 23.0 °C
above_23 = [
    row["timestamp"]
    for row in sensor_data
    if row["temperature_c"] > 23.0 and row["status"] == "OK"
]
print(f"\nTimestamps above 23.0 °C ({len(above_23)} records):")
for ts in above_23:
    print(f"  {ts}")
```

## Solution — Exercise 4: Sensor Configuration Dictionary


```python
# Create
sensor_config = {
    "id": "S1",
    "location": "Northwest column, level 2",
    "sampling_rate_hz": 1.0,
    "temp_range_c": (-20.0, 60.0),
    "active": True
}

# 1. Print whole dict
print("Full config:", sensor_config)

# 2. Print location
print(f"Location: {sensor_config['location']}")

# 3. Safe access
fw = sensor_config.get("firmware_version", "unknown")
print(f"Firmware (before add): {fw}")

# 4. Add firmware_version
sensor_config["firmware_version"] = "2.1.4"

# 5. Update sampling rate
sensor_config["sampling_rate_hz"] = 0.5

# 6. Delete active
del sensor_config["active"]

# 7. Iterate
print("\nUpdated config:")
for key, value in sensor_config.items():
    print(f"  {key:<22}: {value}")
```

## Solution — Exercise 5: Dictionary of Lists


```python
grouped = {"S1": [], "S2": []}

for row in sensor_data:
    if row["status"] == "OK":              # skip ERROR rows
        sid = row["sensor_id"]
        grouped[sid].append(row["temperature_c"])

print("Grouped temperatures:")
for sensor_id, temps in grouped.items():
    n = len(temps)
    avg = sum(temps) / n if n > 0 else 0
    print(f"  {sensor_id}: count={n}, min={min(temps):.1f}, max={max(temps):.1f}, avg={avg:.2f}")
```

---

## Solution — Exercise 6: Summary Report


```python
total = len(sensor_data)
errors = sum(1 for row in sensor_data if row["status"] == "ERROR")
valid = total - errors

summary = {
    "total_records": total,
    "error_records": errors,
    "valid_records": valid,
}

for sensor_id, temps in grouped.items():
    n = len(temps)
    summary[sensor_id] = {
        "count": n,
        "avg_temp_c": round(sum(temps) / n, 2) if n > 0 else None,
        "min_temp_c": min(temps) if n > 0 else None,
        "max_temp_c": max(temps) if n > 0 else None,
    }

# Print
print("=" * 40)
print("Sensor Network Summary Report")
print("=" * 40)
print(f"{'Total records':<22}: {summary['total_records']}")
print(f"{'Error records':<22}: {summary['error_records']}")
print(f"{'Valid records':<22}: {summary['valid_records']}")

for sid in ["S1", "S2"]:
    s = summary[sid]
    print(f"\n  Sensor {sid}:")
    print(f"    {'Valid readings':<20}: {s['count']}")
    print(f"    {'Avg temperature':<20}: {s['avg_temp_c']:.2f} °C")
    print(f"    {'Min temperature':<20}: {s['min_temp_c']:.1f} °C")
    print(f"    {'Max temperature':<20}: {s['max_temp_c']:.1f} °C")
```

## Solution — Stretch Goal: Anomaly Detection


```python
# Collect all valid temperatures
valid_temps = [
    row["temperature_c"]
    for row in sensor_data
    if row["status"] == "OK" and row["temperature_c"] > -100
]

n = len(valid_temps)
mean = sum(valid_temps) / n

# Standard deviation — manual implementation
variance = sum((t - mean) ** 2 for t in valid_temps) / n
sigma = variance ** 0.5  # square root without math.sqrt

print(f"Valid readings : {n}")
print(f"Mean           : {mean:.3f} °C")
print(f"Std deviation  : {sigma:.3f} °C")
print(f"Bounds (2σ)    : [{mean - 2*sigma:.3f}, {mean + 2*sigma:.3f}] °C")

# Find anomalies
anomalies = [
    row for row in sensor_data
    if row["status"] == "OK"
    and abs(row["temperature_c"] - mean) > 2 * sigma
]

print(f"\nAnomalies detected: {len(anomalies)}")
for a in anomalies:
    print(f"  {a['timestamp']} | {a['sensor_id']} | {a['temperature_c']:.1f} °C")

# Anomaly report grouped by sensor
anomaly_report = {"S1": [], "S2": []}
for a in anomalies:
    anomaly_report[a["sensor_id"]].append(a)

print("\nAnomaly report by sensor:")
for sid, records in anomaly_report.items():
    print(f"  {sid}: {len(records)} anomalies")
```

**Discussion**: With this small, well-behaved dataset, the 2σ rule likely detects no anomalies (temperatures cluster tightly around 23 °C). The engineering value of this exercise is the pattern itself: collecting valid data, computing statistics, and flagging outliers. In Chapter 5, when we read the actual CSV file, the exact same logic applies to any size dataset.

The standard deviation formula used here is the **population** standard deviation (dividing by n). The **sample** standard deviation divides by n−1. For large datasets the difference is negligible; for this 9-point dataset it is small but non-zero. In professional code, use `statistics.stdev()` (sample) or `statistics.pstdev()` (population) from Python's standard library.
