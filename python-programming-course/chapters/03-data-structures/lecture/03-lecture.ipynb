## 3.1 Lists

A **list** is an ordered, mutable (changeable) sequence of values. Elements can be of any type, and a single list can mix types, though homogeneous lists are most common in engineering data.

### Creating Lists


````python
# List literals
temperatures = [22.3, 22.8, 23.2, 23.5, 24.0]  # list of floats
sensor_ids = ["S1", "S2", "S3"]                  # list of strings
flags = [True, True, False, True]                 # list of booleans
mixed = [1, "S1", 22.3, True]                     # mixed types (rare in practice)
empty = []                                         # empty list

print(f"temperatures : {temperatures}")
print(f"sensor_ids   : {sensor_ids}")
print(f"Length       : {len(temperatures)}")
````

````python
# Indexing — zero-based, negative indices count from end
print(f"First reading  : {temperatures[0]}")
print(f"Last reading   : {temperatures[-1]}")
print(f"Third reading  : {temperatures[2]}")

# Slicing — same syntax as strings: [start:stop:step], stop is exclusive
first_three = temperatures[0:3]   # or temperatures[:3]
last_two = temperatures[-2:]      # last two elements
every_other = temperatures[::2]   # every 2nd element

print(f"\nFirst three : {first_three}")
print(f"Last two    : {last_two}")
print(f"Every other : {every_other}")
````

````python
# Mutability — you can change elements in place
readings = [22.3, -999.0, 23.2]  # second reading is faulty
print(f"Before: {readings}")
readings[1] = 22.8               # replace the faulty reading
print(f"After : {readings}")
````

### List Methods

Python lists have many built-in methods that modify the list in place:


````python
measurements = [22.3, 22.8, 23.2]

# append() — add one element to the end
measurements.append(23.5)
print(f"After append     : {measurements}")

# insert() — add at a specific position
measurements.insert(0, 21.9)  # insert at index 0 (beginning)
print(f"After insert(0,) : {measurements}")

# remove() — remove first occurrence of a value
measurements.remove(22.3)
print(f"After remove     : {measurements}")

# pop() — remove and return element at index (default: last)
last = measurements.pop()
print(f"Popped value     : {last}")
print(f"After pop        : {measurements}")

# sort() — sort in place (returns None!)
data = [5.0, 2.3, 8.1, 1.4, 6.7]
data.sort()
print(f"\nSorted in place  : {data}")

data.sort(reverse=True)
print(f"Sorted descending: {data}")
````

````python
# sorted() — returns a NEW sorted list; original unchanged
raw = [5.0, 2.3, 8.1, 1.4, 6.7]
ordered = sorted(raw)
print(f"Original        : {raw}")
print(f"sorted() result : {ordered}")
````

````python
# Aggregate functions: min, max, sum
temps = [22.3, 22.8, 23.2, 23.5, 24.0]

print(f"Count   : {len(temps)}")
print(f"Minimum : {min(temps):.1f} °C")
print(f"Maximum : {max(temps):.1f} °C")
print(f"Sum     : {sum(temps):.1f} °C")
print(f"Average : {sum(temps) / len(temps):.2f} °C")
````

````python
# The 'in' operator — membership test
active_sensors = ["S1", "S2", "S4"]
print(f"'S2' in active_sensors: {'S2' in active_sensors}")
print(f"'S3' in active_sensors: {'S3' in active_sensors}")
````

### List Comprehensions

A **list comprehension** is a concise, single-line way to build a new list by applying a transformation or filter to an existing sequence:

```python
[expression for item in iterable if condition]
```

The `if condition` part is optional.


````python
# Example 1: convert a list of Celsius temperatures to Kelvin
temps_c = [22.3, 22.8, 23.2, 23.5, 24.0]
temps_k = [t + 273.15 for t in temps_c]
print(f"Kelvin: {temps_k}")

# Example 2: filter out faulty readings (below -100)
raw = [22.3, -999.0, 23.2, 23.5, -999.0, 24.0]
valid = [t for t in raw if t > -100]
print(f"Valid readings: {valid}")

# Example 3: combine filter and transform
valid_k = [t + 273.15 for t in raw if t > -100]
print(f"Valid in Kelvin: {valid_k}")
````

### Check Your Understanding — 3.1

**Question**: What is the difference between `data.sort()` and `sorted(data)`?

**Answer**: `.sort()` modifies `data` in place and returns `None`. `sorted(data)` returns a new sorted list and leaves `data` unchanged. For small datasets this does not matter; for large datasets, knowing which copy is modified is important for memory management.

**Question**: Given `temps = [22.3, 22.8, 23.2]`, what does `temps[1:3]` return?

**Answer**: `[22.8, 23.2]` — indices 1 and 2. The stop index (3) is exclusive.

## 3.2 Dictionaries

A **dictionary** stores **key-value pairs**. Each key maps to exactly one value. Keys must be unique and immutable (strings and integers are the most common key types).

Dictionaries are the natural Python representation for any data with named fields — a sensor record, a configuration parameter set, a material property table.


````python
# Dictionary literals — {key: value, key: value, ...}
sensor = {
    "id": "S1",
    "temperature_c": 22.3,
    "pressure_kpa": 101.2,
    "status": "OK"
}

print(sensor)
print(f"Type: {type(sensor).__name__}")
print(f"Number of keys: {len(sensor)}")
````

````python
# Accessing values by key
print(f"Sensor ID   : {sensor['id']}")
print(f"Temperature : {sensor['temperature_c']:.1f} °C")
print(f"Status      : {sensor['status']}")

# Safe access with .get() — returns None (or a default) if key is missing
humidity = sensor.get('humidity')           # key doesn't exist → None
humidity_default = sensor.get('humidity', 0.0)  # with default value

print(f"Humidity    : {humidity}")          # None
print(f"Humidity (default): {humidity_default:.1f}")
````

````python
# Adding, updating, and deleting entries
config = {"sampling_rate_hz": 10, "timeout_s": 30}

# Add a new key
config["buffer_size"] = 1024
print(f"After add    : {config}")

# Update an existing key
config["timeout_s"] = 60
print(f"After update : {config}")

# Delete a key
del config["buffer_size"]
print(f"After delete : {config}")
````

````python
# Iterating over a dictionary
material_properties = {
    "Young_modulus_GPa": 200,
    "yield_strength_MPa": 250,
    "density_kg_m3": 7850,
    "Poisson_ratio": 0.3
}

print("Steel properties:")

# Iterate over keys
for key in material_properties:
    print(f"  {key}: {material_properties[key]}")

print()

# Iterate over key-value pairs with .items()
for key, value in material_properties.items():
    print(f"  {key:<25} = {value}")
````

````python
# .keys() and .values()
print(f"Keys   : {list(material_properties.keys())}")
print(f"Values : {list(material_properties.values())}")

# Check for key existence
print(f"'density_kg_m3' in dict: {'density_kg_m3' in material_properties}")
print(f"'colour' in dict        : {'colour' in material_properties}")
````

### Nested Dictionaries and List of Dicts

Complex data structures can be built by nesting. The most common engineering pattern is a **list of dictionaries** — each dictionary represents one record (e.g., one sensor reading row).


````python
# List of dictionaries — represents a table of sensor readings
# This is the same data as in assets/sensor_data.csv

readings = [
    {"timestamp": "2024-01-15 08:00:00", "sensor_id": "S1", "temperature_c": 22.3, "pressure_kpa": 101.2, "status": "OK"},
    {"timestamp": "2024-01-15 08:00:00", "sensor_id": "S2", "temperature_c": 23.1, "pressure_kpa": 101.5, "status": "OK"},
    {"timestamp": "2024-01-15 08:15:00", "sensor_id": "S1", "temperature_c": 22.8, "pressure_kpa": 101.3, "status": "OK"},
    {"timestamp": "2024-01-15 08:30:00", "sensor_id": "S2", "temperature_c": -999.0, "pressure_kpa": 101.6, "status": "ERROR"},
    {"timestamp": "2024-01-15 09:00:00", "sensor_id": "S1", "temperature_c": 24.0, "pressure_kpa": 100.9, "status": "OK"},
]

# Access the second reading
print(f"Second reading: {readings[1]}")

# Access a specific field of a specific reading
print(f"Sensor in 3rd row: {readings[2]['sensor_id']}")
print(f"Temperature in 3rd row: {readings[2]['temperature_c']} °C")
````

````python
# Filter: collect only valid S1 readings
s1_valid = [
    row
    for row in readings
    if row["sensor_id"] == "S1" and row["status"] == "OK"
]

s1_temps = [row["temperature_c"] for row in s1_valid]

print(f"S1 valid readings : {len(s1_valid)}")
print(f"S1 temperatures   : {s1_temps}")
print(f"S1 average temp   : {sum(s1_temps) / len(s1_temps):.2f} °C")
````

````python
# Print a formatted table from the list of dicts
print(f"{'Timestamp':<22} {'Sensor':<8} {'Temp (°C)':>10} {'Pressure (kPa)':>16} {'Status':<8}")
print("-" * 68)

for row in readings:
    temp_str = f"{row['temperature_c']:>10.1f}" if row["status"] == "OK" else f"{'ERR':>10}"
    print(f"{row['timestamp']:<22} {row['sensor_id']:<8} {temp_str} {row['pressure_kpa']:>16.1f} {row['status']:<8}")
````

### Check Your Understanding — 3.2

**Question**: You have the dictionary `d = {'a': 1, 'b': 2}`. What happens if you do `d['c']`? What should you use instead?

**Answer**: `d['c']` raises a `KeyError` because `'c'` does not exist. Use `d.get('c')` (returns `None`) or `d.get('c', 0)` (returns `0` as a default). If you want to raise an informative error on purpose, `d['c']` is fine — but for production code where key presence is uncertain, always use `.get()`.

## 3.3 Choosing Between Lists and Dictionaries

| Question | Use a list | Use a dictionary |
|---|---|---|
| Does order matter? | Yes | Not necessarily |
| Do you access by position? | Yes (`data[2]`) | No |
| Do you access by a name/label? | No | Yes (`data['temperature']`) |
| Is the data uniform? | Usually | Mixed types common |
| Classic example | 1000 temperature readings | One sensor's properties |

For a dataset of many records (rows), the **list of dicts** pattern combines both: the list provides ordered access by row number; each dict provides named access to fields within a row.

## Chapter 3 Summary

1. **Lists** are ordered, mutable sequences. Key operations: indexing (`[i]`), slicing (`[a:b]`), `.append()`, `.remove()`, `.sort()`, `sorted()`, `len()`, `min()`, `max()`, `sum()`, `in`.
2. **List comprehensions** concisely build filtered or transformed lists: `[expr for x in seq if cond]`.
3. **Dictionaries** store key-value pairs. Key operations: `d[key]`, `d.get(key, default)`, `d[key] = val`, `del d[key]`, `.keys()`, `.values()`, `.items()`, `in` for key membership.
4. **List of dicts** is the standard pattern for tabular engineering data.
