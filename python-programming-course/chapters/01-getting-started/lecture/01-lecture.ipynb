## What is Python?

Python is a **general-purpose, interpreted, high-level programming language** created by Guido van Rossum and first released in 1991. Its design philosophy emphasises code readability — Python programs are often described as 'executable pseudocode' because the syntax closely mirrors plain English.

Key characteristics relevant to engineers:

- **Interpreted**: Python executes code line by line without a separate compilation step. You get immediate feedback, which makes exploration and debugging fast.
- **Dynamically typed**: You do not declare variable types. Python determines them at runtime.
- **Open-source and free**: No licensing cost. A large community produces thousands of freely available libraries.
- **Cross-platform**: The same code runs on Windows, macOS, and Linux with minimal or no changes.

### Python 2 vs Python 3

Python 3 is the current and only supported major version. Python 2 reached **end-of-life on 1 January 2020** and should never be used for new projects. All course materials target **Python 3.12**.

The most visible difference is `print`: in Python 2 it was a statement (`print 'hello'`); in Python 3 it is a function (`print('hello')`).

### Why Python in Engineering?

Engineering curricula worldwide have adopted Python because:
1. It is used in industry for automation, data analysis, and simulation.
2. Libraries such as NumPy, SciPy, and Pandas extend it into scientific computing.
3. The syntax overhead is low — you can express an algorithm clearly without writing dozens of boilerplate lines.

## 1.1 The Python Command-Line REPL

REPL stands for **Read–Eval–Print Loop**. It is the interactive Python shell that:

1. **Reads** an expression or statement you type.
2. **Evaluates** it immediately.
3. **Prints** the result.
4. **Loops** back to wait for the next input.

You launch it by opening a terminal and typing:

```
python
```

or on macOS/Linux:

```
python3
```

The shell greets you with the Python prompt `>>>`:

```
Python 3.12.0 (main, Oct  2 2023, 12:00:00)
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

To exit the REPL, type `exit()` or press `Ctrl+D` (Linux/macOS) / `Ctrl+Z` then Enter (Windows).

The cells below simulate REPL interactions. In Jupyter, each code cell behaves similarly: type code and press `Shift+Enter` to execute.


````python
# The simplest possible expression — Python evaluates 2 + 2 and prints 4
2 + 2
````

````python
# String expression — the REPL prints the string representation
'Hello, engineering world!'
````

````python
# Python handles floating-point arithmetic naturally
# Example: compute the gravitational force on a 75 kg person
mass_kg = 75.0
GRAVITY = 9.81  # m/s², standard gravitational acceleration
weight_n = mass_kg * GRAVITY
print(f"Weight: {weight_n:.2f} N")
````

````python
# The REPL is useful for checking Python version information
import sys
print(f"Python version: {sys.version}")
print(f"Version info  : {sys.version_info.major}.{sys.version_info.minor}")
````

## 1.2 IEP — Interactive Editor for Python (Historical Context)

The official course syllabus mentions **IEP (Interactive Editor for Python)**. IEP was an open-source, cross-platform Python IDE designed for scientific and interactive use. It featured:

- A code editor pane and an interactive Python shell side by side.
- A variable explorer showing live values of variables.
- Shell workspaces for different Python environments.

IEP was developed primarily between 2010 and 2015. Its last stable release was version 3.6 (approximately 2015), and the project has not been actively maintained since then.

### Why IEP Matters in this Context

Understanding IEP is worthwhile because it introduced concepts that are now standard in scientific Python IDEs:

| IEP concept | Modern equivalent |
|---|---|
| Interactive shell + editor | JupyterLab, VS Code |
| Variable explorer | JupyterLab variable inspector, Spyder |
| Kernel-based execution | Jupyter kernels |

### Current Recommendation

For this course, use one of these proven modern tools:

1. **JupyterLab** — for all notebook-based work (all chapter labs).
2. **IDLE** — for simple script editing and REPL exploration (Chapter 1 focus).
3. **VS Code with Python/Jupyter extensions** — recommended for students who want a full IDE experience.

### What 'PATH' means

The PATH is an environment variable that lists directories the operating system searches when you type a command. When Python is on the PATH, the OS finds the `python.exe` executable from any directory. Without it, you would have to type the full path every time: `C:\Python312\python.exe script.py`.

The code cell below checks whether the Python executable is accessible and prints its location:


````python
import sys
import os

print(f"Python executable : {sys.executable}")
print(f"Python version    : {sys.version.split()[0]}")
print(f"Platform          : {sys.platform}")
print(f"Working directory : {os.getcwd()}")
````

## 1.4 IDLE — Python's Built-In GUI

**IDLE** (Integrated Development and Learning Environment) ships with every standard Python installation. It provides:

1. **Shell window** — an interactive Python REPL with colour syntax highlighting.
2. **Script editor** — File → New File creates a `.py` file with a full editor; press F5 to run it.

IDLE is intentionally simple — it is designed to be approachable for beginners without overwhelming menus. For a first course in Python, IDLE is perfectly adequate.

### Launching IDLE

| Platform | Method |
|---|---|
| Windows | Search "IDLE" in Start menu |
| macOS | Applications → Python 3.12 → IDLE |
| Linux | Run `idle3` in terminal |

You can also launch it from the terminal:

```
python -m idlelib.idle
```

### IDLE Shell vs Script Editor

The **Shell window** is the REPL — ideal for quick experiments. The **script editor** is for writing multi-line programs that you save and run repeatedly.

Workflow in IDLE:
1. Open IDLE → you see the shell (`>>>` prompt).
2. File → New File → a blank editor opens.
3. Write your program.
4. File → Save As → `hello.py`.
5. Run → Run Module (or press **F5**) → output appears in the shell.

**Question**: You write `x = 5` in the IDLE shell. Then you close IDLE and reopen it. Is `x` still defined?

## 1.5 Jupyter Notebooks — Your Primary Lab Environment

All course lab work is done in **Jupyter Notebooks** (`.ipynb` files). A notebook is a document that interleaves:

### The Kernel

Behind every notebook is a **kernel** — a running Python process. Variables defined in one cell persist in memory for subsequent cells within the same session. If the kernel is restarted (Kernel → Restart), all variables are cleared.

> **Engineering analogy**: Think of the kernel as a calculator that remembers its internal registers between presses of the `=` key. Restarting the kernel is like pressing `AC` (all clear).


````python
# Demonstrating kernel state persistence across cells:
# Define a variable in this cell...
beam_span_m = 8.0  # metres
````

````python
# ...and use it in this cell — it is still in memory.
load_per_m = 5.0  # kN/m, uniformly distributed load
total_load_kn = load_per_m * beam_span_m
print(f"Beam span       : {beam_span_m} m")
print(f"UDL             : {load_per_m} kN/m")
print(f"Total load      : {total_load_kn} kN")
````
