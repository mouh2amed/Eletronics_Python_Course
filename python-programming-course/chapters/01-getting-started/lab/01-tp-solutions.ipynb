## Solution — Exercise 1: Verify Your Python Installation

The `sys` module provides access to Python interpreter information. `sys.version` is a string containing the full version; `sys.version_info` is a named tuple with individual components (`major`, `minor`, `micro`). Using `sys.version_info` components is more robust than parsing the string, because it does not depend on the exact string format.


````python
import sys

print(f"Python version : {sys.version.split()[0]}")
print(f"Major version  : {sys.version_info.major}")
print(f"Minor version  : {sys.version_info.minor}")
print(f"Executable     : {sys.executable}")
````

**Note**: `sys.version` contains more text than just the version number (compilation date, compiler info). Calling `.split()[0]` extracts only the first whitespace-separated token, which is the version number string `'3.12.x'`.

**Alternative**: `platform.python_version()` returns a clean version string and is sometimes preferred for display purposes:
```python
import platform
print(platform.python_version())  # '3.12.0'
```

## Solution — Exercise 2: Working Directory Awareness

`os.getcwd()` returns the **current working directory** as a string. This is the directory Python searches by default when you use relative file paths like `open('data.csv')`. Always verify the CWD when debugging file-not-found errors.


````python
import os

cwd = os.getcwd()
print(f"Current working directory: {cwd}")
````

**Engineering context**: In a data processing script that reads `sensor_data.csv`, if that file is in `assets/` but your CWD is the project root, you must use the path `'assets/sensor_data.csv'`. If your CWD is already inside `assets/`, you only need `'sensor_data.csv'`. Being aware of the CWD prevents one of the most common beginner errors.

## Solution — Exercise 3: Your First print() Statements

The key points here are: (1) using multiple `print()` calls for multiple lines, (2) using an f-string to embed the actual Python version dynamically rather than hard-coding it.


````python
import sys

version_str = f"{sys.version_info.major}.{sys.version_info.minor}"
separator = "=" * 50

print(separator)
print("  Algerian Engineering Consultancy — Python Environment")
print(f"  Python {version_str} | Ready for use")
print("  Run 'jupyter lab' to open the lab notebooks.")
print(separator)
````

**Style note**: Building the separator string as `"=" * 50` is cleaner than typing 50 equals signs. This is Python's string repetition operator. We also separate the version computation from the print calls for clarity — a small but meaningful readability improvement.

## Solution — Exercise 5: IDLE Workflow (Written Answers)

1. The keyboard shortcut to run a script in IDLE is: **F5**

2. To open the IDLE script editor from the shell window, use: **File → New File**

3. The IDLE shell displays the Python prompt: **`>>>`**

4. If you define `x = 100` in the IDLE shell and then close IDLE, `x` will be **no longer defined** when you reopen IDLE. Variables exist only in the running Python process (the kernel/session). Closing IDLE terminates that process.

5. Running a script with F5 vs typing in the shell: F5 runs the entire saved `.py` file in sequence as a fresh execution. Typing directly in the shell executes one statement at a time in the current interactive session, which retains state from previous commands.

## Solution — Stretch Goal: Environment Report Script

The `:<25` format specifier in an f-string left-aligns the text within a field of width 25 characters, producing neatly aligned columns. `os.listdir('.')` returns a list of filenames in the current directory.


````python
import sys
import os

version_str = f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}"
contents = os.listdir('.')
contents.sort()  # Sort alphabetically for consistent output

print("=" * 45)
print("=== Python Environment Report ===")
print("=" * 45)
print(f"{'Python version':<25}: {version_str}")
print(f"{'Platform':<25}: {sys.platform}")
print(f"{'Executable':<25}: {sys.executable}")
print(f"{'Working directory':<25}: {os.getcwd()}")
print(f"{'Directory contents':<25}:")
for item in contents:
    print(f"  - {item}")
print(f"{'Total items':<25}: {len(contents)}")
print("=" * 45)
````

**Alternative approach**: The `platform` module provides cleaner access to OS information:
```python
import platform
print(platform.system())    # 'Windows', 'Linux', 'Darwin'
print(platform.node())      # hostname
print(platform.python_version())  # '3.12.0'
```

**Engineering note**: A script like this is useful as a sanity-check header at the top of data-processing notebooks. Running it confirms that the analysis was performed with the expected Python version and from the correct directory — important for reproducibility.
