## 4.1 Built-In Functions

Python ships with a large collection of functions available without any import. The ones most relevant to engineering computation are:

| Function | Purpose | Example |
|---|---|---|
| `print(x)` | Output to screen | `print("hello")` |
| `len(s)` | Length of sequence | `len([1,2,3])` → `3` |
| `range(n)` | Integer range | `range(0, 5)` → `0,1,2,3,4` |
| `type(x)` | Type of object | `type(3.14)` → `<class 'float'>` |
| `int(x)` | Convert to integer | `int("42")` → `42` |
| `float(x)` | Convert to float | `float("3.14")` → `3.14` |
| `str(x)` | Convert to string | `str(42)` → `"42"` |
| `abs(x)` | Absolute value | `abs(-5.3)` → `5.3` |
| `round(x, n)` | Round to n decimals | `round(3.14159, 2)` → `3.14` |
| `min(iterable)` | Minimum value | `min([3,1,2])` → `1` |
| `max(iterable)` | Maximum value | `max([3,1,2])` → `3` |
| `sum(iterable)` | Sum of values | `sum([1,2,3])` → `6` |
| `sorted(iterable)` | Return sorted copy | `sorted([3,1,2])` → `[1,2,3]` |
| `enumerate(s)` | Index + value pairs | `enumerate(['a','b'])` → `(0,'a'),(1,'b')` |
| `zip(a, b)` | Pair elements | `zip([1,2],[3,4])` → `(1,3),(2,4)` |


````python
# Demonstrating built-in functions with engineering context
force_readings = [120.5, 135.2, 129.8, 141.0, 118.3, 133.7]

print(f"Count    : {len(force_readings)}")
print(f"Min      : {min(force_readings):.1f} N")
print(f"Max      : {max(force_readings):.1f} N")
print(f"Sum      : {sum(force_readings):.1f} N")
print(f"Average  : {sum(force_readings)/len(force_readings):.2f} N")
print(f"Sorted   : {sorted(force_readings)}")
print(f"Abs(-5.5): {abs(-5.5)}")
print(f"Round    : {round(129.8347, 2)}")
````

````python
# help() — access built-in documentation from the REPL or notebook
# help(round)   # uncomment to see the full docstring

# zip() — useful for iterating parallel lists
sensor_ids = ["S1", "S2", "S3"]
temperatures = [22.3, 23.1, 22.8]

for sid, temp in zip(sensor_ids, temperatures):
    print(f"{sid}: {temp:.1f} °C")
````

## 4.2 Simple Parameterless Functions

The `def` keyword begins a function definition. The colon ends the header; the indented block is the body.

```python
def function_name():
    """Docstring describing what the function does."""
    # function body
    return value  # optional
```

A function that does not have a `return` statement returns `None` implicitly.


````python
# Parameterless function — produces output
def print_separator():
    """Print a standard separator line for reports."""
    print("=" * 50)


# Parameterless function — returns a value
def gravity():
    """Return standard gravitational acceleration in m/s²."""
    return 9.81


# Calling the functions
print_separator()
print("Beam Design Report")
print_separator()

g = gravity()
print(f"g = {g} m/s²")
````

````python
# Demonstrating the difference between print and return
def area_printer():
    """Computes and PRINTS an area — the result is not usable by the caller."""
    area = 3.14 * 0.5 ** 2
    print(f"Area = {area:.4f} m²")


def area_returner():
    """Computes and RETURNS an area — the caller can use the result."""
    area = 3.14 * 0.5 ** 2
    return area


# area_printer: result is gone after the print
result_a = area_printer()   # prints, but result_a is None
print(f"result_a = {result_a}")   # None

# area_returner: result is available
result_b = area_returner()
print(f"result_b = {result_b:.4f} m²")   # usable value
````

## 4.3 Functions with Parameters

Parameters are declared inside the parentheses. Arguments are the actual values passed when calling the function.


````python
def celsius_to_kelvin(temp_c):
    """Convert a temperature from Celsius to Kelvin."""
    return temp_c + 273.15


def compute_stress(force_n, area_m2):
    """
    Compute normal stress.

    Parameters
    ----------
    force_n : float
        Applied force in Newtons.
    area_m2 : float
        Cross-sectional area in square metres.

    Returns
    -------
    float
        Normal stress in Pascals.
    """
    return force_n / area_m2


# Calling with positional arguments
print(f"22.5 °C = {celsius_to_kelvin(22.5):.2f} K")

force = 50000   # N
area = 0.002    # m²
stress = compute_stress(force, area)
print(f"Stress = {stress / 1e6:.2f} MPa")
````

````python
# Multiple return values — Python returns them as a tuple
def beam_reactions(total_load_n, span_m):
    """
    Compute reactions at both supports of a simply-supported beam
    with a central point load.
    Returns (left_reaction_n, right_reaction_n).
    """
    # By symmetry, each support carries half the load.
    reaction = total_load_n / 2
    return reaction, reaction  # returns a tuple


R_left, R_right = beam_reactions(10000, 6.0)  # tuple unpacking
print(f"Left reaction  : {R_left / 1000:.1f} kN")
print(f"Right reaction : {R_right / 1000:.1f} kN")
````

## 4.4 Default Parameter Values

A parameter with a default value becomes optional in the function call. Parameters with defaults must come **after** all required parameters.


````python
def format_reading(value, unit="°C", precision=2):
    """
    Format a sensor reading with unit and decimal precision.

    Parameters
    ----------
    value : float
        The measured value.
    unit : str, optional
        Unit string (default: '°C').
    precision : int, optional
        Number of decimal places (default: 2).
    """
    return f"{value:.{precision}f} {unit}"


print(format_reading(22.386))                     # uses all defaults
print(format_reading(101.325, unit="kPa"))         # override unit only
print(format_reading(22.386, precision=1))         # override precision only
print(format_reading(9.81, unit="m/s²", precision=3))  # override both
````

````python
# WARNING: Mutable default argument anti-pattern
# NEVER use a mutable object (list, dict) as a default value.

# WRONG
def append_reading_wrong(value, readings=[]):
    readings.append(value)
    return readings

print(append_reading_wrong(1.0))  # [1.0]   — OK
print(append_reading_wrong(2.0))  # [1.0, 2.0] — WRONG: the list is shared!

# CORRECT — use None as default and create inside the function
def append_reading_correct(value, readings=None):
    if readings is None:
        readings = []
    readings.append(value)
    return readings

print(append_reading_correct(1.0))  # [1.0]
print(append_reading_correct(2.0))  # [2.0]  — each call gets a fresh list
````

## 4.5 Assigning a Function to a Variable

In Python, functions are **first-class objects** — they can be assigned to variables, stored in lists, and passed as arguments to other functions.


````python
def celsius_to_fahrenheit(t):
    """Convert Celsius to Fahrenheit."""
    return t * 9 / 5 + 32

def celsius_to_kelvin(t):
    """Convert Celsius to Kelvin."""
    return t + 273.15


# Assign a function to a variable — no parentheses!
convert = celsius_to_fahrenheit
print(f"100 °C via convert() = {convert(100):.1f}")  # calls celsius_to_fahrenheit

# Dynamically choose which conversion to apply
conversions = {
    "F": celsius_to_fahrenheit,
    "K": celsius_to_kelvin
}

for unit, func in conversions.items():
    print(f"100 °C → {func(100):.2f} {unit}")
````

````python
# Passing a function as an argument to another function
def apply_conversion(temperatures, converter_func):
    """
    Apply a conversion function to each temperature in a list.

    Parameters
    ----------
    temperatures : list of float
        Source temperatures.
    converter_func : callable
        A function that takes one float and returns one float.

    Returns
    -------
    list of float
        Converted temperatures.
    """
    return [converter_func(t) for t in temperatures]


temps_c = [0.0, 20.0, 37.0, 100.0]
temps_f = apply_conversion(temps_c, celsius_to_fahrenheit)
temps_k = apply_conversion(temps_c, celsius_to_kelvin)

print("°C →  °F     K")
for c, f, k in zip(temps_c, temps_f, temps_k):
    print(f"{c:5.1f} → {f:6.1f}  {k:6.2f}")
````

## 4.6 Variadic Functions: *args and **kwargs

Sometimes you do not know in advance how many arguments a function will receive. Python provides two special parameter syntaxes:

- `*args` — collects any number of **positional** arguments into a **tuple**.
- `**kwargs` — collects any number of **keyword** arguments into a **dictionary**.


````python
# *args — variable number of positional arguments
def average(*values):
    """
    Return the average of any number of numeric values.

    Parameters
    ----------
    *values : float
        Any number of numeric arguments.
    """
    if not values:
        return 0.0
    return sum(values) / len(values)


print(average(22.3))                          # 1 argument
print(average(22.3, 22.8, 23.2))              # 3 arguments
print(average(22.3, 22.8, 23.2, 23.5, 24.0)) # 5 arguments
````

````python
# Combining fixed parameters with *args
def sensor_report(sensor_id, *readings):
    """
    Print a report for a sensor with any number of readings.

    Parameters
    ----------
    sensor_id : str
        Sensor identifier.
    *readings : float
        Any number of temperature readings.
    """
    print(f"Sensor {sensor_id} — {len(readings)} reading(s)")
    if readings:
        print(f"  Values  : {list(readings)}")
        print(f"  Average : {sum(readings)/len(readings):.2f} °C")


sensor_report("S1", 22.3, 22.8, 23.2, 23.5)
print()
sensor_report("S2")   # no readings provided
````

````python
# **kwargs — variable number of keyword arguments
def create_material(**properties):
    """
    Create a material properties dictionary from keyword arguments.

    Parameters
    ----------
    **properties : any
        Arbitrary material properties as keyword arguments.
    """
    print(f"Material with {len(properties)} properties:")
    for name, value in properties.items():
        print(f"  {name}: {value}")
    return properties


steel = create_material(
    name="Structural steel S355",
    youngs_modulus_gpa=200,
    yield_strength_mpa=355,
    density_kg_m3=7850
)
````

## 4.7 Mutable vs Immutable Argument Passing

Python passes arguments by **object reference**. What this means in practice depends on whether the argument is mutable or immutable:

- **Immutable** (int, float, str, tuple): the function receives a reference, but it cannot change the original because any assignment creates a new object.
- **Mutable** (list, dict): the function receives a reference to the same object. If the function modifies the object in-place (e.g., `.append()`), the change is visible to the caller.


````python
# Immutable: modifying inside function does NOT affect caller
def add_offset_wrong(value, offset):
    """Attempt to modify an int in place — does NOT work."""
    value = value + offset  # creates a NEW int object; original unchanged
    print(f"  Inside function: value = {value}")


x = 10
print(f"Before: x = {x}")
add_offset_wrong(x, 5)
print(f"After : x = {x}")   # Still 10
````

````python
# Mutable: modifying a list IN PLACE affects the caller
def remove_faults_inplace(readings):
    """Remove fault values from the list in place (modifies caller's list!)."""
    i = 0
    while i < len(readings):
        if readings[i] < -100:
            readings.pop(i)
        else:
            i += 1


data = [22.3, -999.0, 23.1, -999.0, 23.5]
print(f"Before: {data}")
remove_faults_inplace(data)
print(f"After : {data}")   # Original list was modified!
````

````python
# Better practice: return a NEW list instead of modifying in place
def remove_faults(readings):
    """Return a new list with fault values removed. Original is untouched."""
    return [t for t in readings if t > -100]


data = [22.3, -999.0, 23.1, -999.0, 23.5]
cleaned = remove_faults(data)
print(f"Original : {data}")
print(f"Cleaned  : {cleaned}")
````

## 4.8 Local and Global Scope

Python resolves variable names using the **LEGB** rule, in this order:
1. **L**ocal — variables defined inside the current function.
2. **E**nclosing — variables in enclosing functions (relevant for nested functions).
3. **G**lobal — variables defined at the module (top) level.
4. **B**uilt-in — Python's built-in names (`len`, `print`, etc.).

A function can **read** global variables, but cannot **reassign** them without the `global` keyword.


````python
# Global constant — readable inside functions without any declaration
GRAVITY = 9.81  # global constant

def compute_weight(mass_kg):
    """Compute weight in Newtons using the global GRAVITY constant."""
    return mass_kg * GRAVITY  # reads GRAVITY from global scope — fine


print(f"Weight of 70 kg = {compute_weight(70):.2f} N")
````

````python
# Local variable shadows global — the global is unchanged
x = 100  # global

def set_x():
    x = 42  # local — creates a NEW variable; the global x = 100 is untouched
    print(f"Inside function: x = {x}")

set_x()
print(f"Global after call: x = {x}")  # still 100
````

````python
# The global keyword — use sparingly; prefer passing arguments
reading_count = 0  # global counter

def record_reading():
    """Increment the global reading counter."""
    global reading_count      # tells Python: use the global variable
    reading_count += 1


record_reading()
record_reading()
record_reading()
print(f"Total readings recorded: {reading_count}")  # 3
````

**Engineering guideline**: Use global variables only for module-level constants (e.g., `GRAVITY = 9.81`). For anything that changes, prefer passing the value as an argument and returning the updated value.

## 4.9 Lambda Functions

A **lambda** is an anonymous (nameless) function defined in a single expression:

```python
lambda parameters: expression
```

Lambdas are most useful when you need a short function in a context that expects a callable — most commonly as the `key` argument to `sorted()`, `min()`, or `max()`.


````python
# Simple lambda functions
to_kelvin = lambda t: t + 273.15
to_fahrenheit = lambda t: t * 9 / 5 + 32

print(f"0 °C in K : {to_kelvin(0):.2f}")
print(f"100 °C in F: {to_fahrenheit(100):.1f}")
````

````python
# Lambda as key function for sorting
readings = [
    {"sensor_id": "S2", "temperature_c": 23.1},
    {"sensor_id": "S1", "temperature_c": 22.3},
    {"sensor_id": "S3", "temperature_c": 24.0},
]

# Sort by temperature
by_temp = sorted(readings, key=lambda r: r["temperature_c"])
print("Sorted by temperature:")
for r in by_temp:
    print(f"  {r['sensor_id']}: {r['temperature_c']:.1f} °C")

# Sort by sensor ID
by_id = sorted(readings, key=lambda r: r["sensor_id"])
print("\nSorted by sensor ID:")
for r in by_id:
    print(f"  {r['sensor_id']}: {r['temperature_c']:.1f} °C")
````

````python
# Lambda with filter() and map() — functional programming style
temperatures = [22.3, -999.0, 23.1, 22.8, -999.0, 24.0]

# filter() returns only elements where the function is True
valid = list(filter(lambda t: t > -100, temperatures))
print(f"Valid : {valid}")

# map() transforms each element
in_kelvin = list(map(lambda t: t + 273.15, valid))
print(f"Kelvin: {[round(k, 2) for k in in_kelvin]}")

# Equivalent list comprehensions (often more readable)
valid_lc = [t for t in temperatures if t > -100]
in_kelvin_lc = [t + 273.15 for t in valid_lc]
````

### Check Your Understanding — 4.9

**Question**: When should you use a lambda instead of a named function?

**Answer**: Use a lambda when the function is short (one expression), used in exactly one place, and its meaning is clear from context — most commonly as the `key=` argument to `sorted()`. For anything that needs an `if/else` block, a loop, or is used in more than one place, define a named function with a docstring.

## Chapter 4 Summary

1. **Built-in functions** — `len`, `range`, `type`, `abs`, `round`, `min`, `max`, `sum`, `sorted`, `enumerate`, `zip` are available without any import.
2. **Parameterless functions** — defined with `def name():`; distinguish functions that print from functions that return.
3. **Functions with parameters** — positional arguments; multiple returns as tuples; always write a docstring.
4. **Default parameters** — make arguments optional; never use mutable objects as defaults.
5. **Functions as variables** — assign without `()`; pass as arguments; store in dictionaries.
6. **`*args` / `**kwargs`** — accept any number of positional or keyword arguments.
7. **Mutable vs immutable** — ints/strings/floats are effectively passed by value; lists/dicts can be modified in-place by the function.
8. **Scope (LEGB)** — local beats global; use `global` sparingly; prefer arguments over globals.
9. **Lambda** — anonymous single-expression function; best used as a `key=` argument.
