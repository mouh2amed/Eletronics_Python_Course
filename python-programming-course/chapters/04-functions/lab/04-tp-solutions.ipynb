## Solution — Exercise 1: Built-in Functions Statistics


```python
stress_mpa = [187.3, 193.1, 185.7, 201.4, 196.2, 189.8, 204.1, 192.6, 188.5, 197.9]

count = len(stress_mpa)
minimum = min(stress_mpa)
maximum = max(stress_mpa)
total = sum(stress_mpa)
mean = total / count
closest = min(stress_mpa, key=lambda x: abs(x - mean))
three_smallest = sorted(stress_mpa)[:3]

print(f"Count         : {count}")
print(f"Minimum       : {minimum:.1f} MPa")
print(f"Maximum       : {maximum:.1f} MPa")
print(f"Sum           : {total:.1f} MPa")
print(f"Mean          : {mean:.2f} MPa")
print(f"Closest to mean: {closest:.1f} MPa")
print(f"Three smallest: {three_smallest}")
```

## Solution — Exercise 2: Parameterless Functions


```python
def print_report_header():
    """Print a standardised report header with separator lines."""
    separator = "=" * 50
    print(separator)
    print("  Engineering Calculations — ELNI 5.5")
    print(separator)


def gravitational_acceleration():
    """Return standard gravitational acceleration in m/s²."""
    return 9.81


print_report_header()

g = gravitational_acceleration()
mass_kg = 500.0
weight_n = mass_kg * g
print(f"g            = {g} m/s²")
print(f"Weight (500 kg) = {weight_n:.1f} N = {weight_n/1000:.3f} kN")

# If gravitational_acceleration() used print(9.81) instead of return 9.81:
# g = gravitational_acceleration()  would print '9.81' as a side effect,
# but g would be assigned None. Then 'mass_kg * g' would raise TypeError:
# "unsupported operand type(s) for *: 'float' and 'NoneType'".
```

## Solution — Exercise 3: Functions with Parameters


```python
def moment_of_inertia_rect(width_m, height_m):
    """
    Compute the second moment of area of a rectangular section.

    Formula: I = b × h³ / 12

    Parameters
    ----------
    width_m : float
        Section width in metres.
    height_m : float
        Section height in metres.

    Returns
    -------
    float
        Second moment of area in m⁴.
    """
    return width_m * height_m ** 3 / 12


def reynolds_number(velocity_ms, diameter_m, density_kg_m3, viscosity_pa_s):
    """
    Compute the Reynolds number for pipe flow.

    Formula: Re = (v × d × ρ) / μ

    Parameters
    ----------
    velocity_ms : float
        Flow velocity in m/s.
    diameter_m : float
        Pipe internal diameter in metres.
    density_kg_m3 : float
        Fluid density in kg/m³.
    viscosity_pa_s : float
        Dynamic viscosity in Pa·s.

    Returns
    -------
    float
        Reynolds number (dimensionless).
    """
    return (velocity_ms * diameter_m * density_kg_m3) / viscosity_pa_s


def classify_flow(re):
    """
    Classify pipe flow regime based on Reynolds number.

    Parameters
    ----------
    re : float
        Reynolds number.

    Returns
    -------
    str
        One of 'Laminar', 'Transitional', or 'Turbulent'.
    """
    if re < 2300:
        return "Laminar"
    elif re <= 4000:
        return "Transitional"
    else:
        return "Turbulent"


# Test 1: Rectangular section 200mm × 400mm
I = moment_of_inertia_rect(0.200, 0.400)
print(f"Moment of inertia (200×400 mm): {I:.6e} m⁴")

# Test 2: Water at 20°C, d=50mm, v=0.5 m/s
Re = reynolds_number(0.5, 0.05, 998.0, 1.002e-3)
print(f"Reynolds number: {Re:,.0f} → {classify_flow(Re)}")
```

## Solution — Exercise 4: Default Parameters


```python
def format_engineering_value(value, unit, precision=3, width=12):
    """
    Format an engineering value with unit, precision, and column width.

    Parameters
    ----------
    value : float
        Numeric value to format.
    unit : str
        Unit string (e.g., '°C', 'kPa').
    precision : int, optional
        Decimal places (default: 3).
    width : int, optional
        Total field width for the number (default: 12).

    Returns
    -------
    str
        Formatted string like '  22.300 °C'.
    """
    return f"{value:{width}.{precision}f} {unit}"


print(format_engineering_value(22.3, "°C"))             # all defaults
print(format_engineering_value(101.325, "kPa", precision=2))  # 2 decimal places
print(format_engineering_value(9.81, "m/s²", precision=4, width=8))


def beam_deflection(force_n, span_m, E=200e9, I=8.33e-5):
    """
    Compute midspan deflection of a simply-supported beam under central point load.

    Formula: δ = F × L³ / (48 × E × I)

    Parameters
    ----------
    force_n : float
        Applied central point load in Newtons.
    span_m : float
        Beam span in metres.
    E : float, optional
        Young's modulus in Pascals (default: 200 GPa — structural steel).
    I : float, optional
        Second moment of area in m⁴ (default: 8.33e-5 m⁴).

    Returns
    -------
    float
        Midspan deflection in metres.
    """
    return force_n * span_m ** 3 / (48 * E * I)


# 1. Steel beam (defaults)
d1 = beam_deflection(50000, 6.0)
print(f"\nSteel beam   : δ = {d1*1000:.3f} mm")

# 2. Concrete beam (override E)
d2 = beam_deflection(50000, 6.0, E=30e9)
print(f"Concrete beam: δ = {d2*1000:.3f} mm")

# 3. Aluminium beam (override both E and I)
d3 = beam_deflection(50000, 6.0, E=70e9, I=2.5e-5)
print(f"Aluminium    : δ = {d3*1000:.3f} mm")
```

## Solution — Exercise 5: Function Variables


```python
def to_fahrenheit(c):
    """Convert Celsius to Fahrenheit."""
    return c * 9 / 5 + 32


def to_kelvin(c):
    """Convert Celsius to Kelvin."""
    return c + 273.15


def to_rankine(c):
    """Convert Celsius to Rankine (K × 9/5)."""
    return (c + 273.15) * 9 / 5


converters = {
    "F": to_fahrenheit,
    "K": to_kelvin,
    "Ra": to_rankine
}


def batch_convert(temp_c, target_units):
    """
    Convert a Celsius temperature to multiple target units.

    Parameters
    ----------
    temp_c : float
        Temperature in Celsius.
    target_units : list of str
        List of unit codes ('F', 'K', 'Ra').

    Returns
    -------
    dict
        Mapping from unit string to converted value.
    """
    result = {}
    for unit in target_units:
        if unit in converters:
            result[unit] = converters[unit](temp_c)
        else:
            result[unit] = None  # unknown unit
    return result


results = batch_convert(100.0, ["F", "K", "Ra"])
print("100 °C converted:")
for unit, value in results.items():
    print(f"  {unit:3s}: {value:.2f}")
```

## Solution — Exercise 6: *args and **kwargs


```python
def channel_statistics(*channel_readings):
    """
    Compute basic statistics for any number of channel readings.

    Parameters
    ----------
    *channel_readings : float
        Any number of numeric readings.

    Returns
    -------
    dict with keys 'count', 'mean', 'min', 'max', or None values if no readings.
    """
    if not channel_readings:
        return {"count": 0, "mean": None, "min": None, "max": None}
    n = len(channel_readings)
    return {
        "count": n,
        "mean": sum(channel_readings) / n,
        "min": min(channel_readings),
        "max": max(channel_readings)
    }


print(channel_statistics())                       # no args
print(channel_statistics(22.3))                   # 1 arg
print(channel_statistics(22.3, 22.8, 23.2, 24.0, 23.5))  # 5 args


def build_sensor_record(sensor_id, timestamp, **measurements):
    """
    Build a sensor record dictionary from ID, timestamp, and measurements.

    Parameters
    ----------
    sensor_id : str
        Sensor identifier.
    timestamp : str
        ISO-format timestamp string.
    **measurements : float
        Any number of named measurement values.

    Returns
    -------
    dict
        Unified record dictionary.
    """
    record = {"sensor_id": sensor_id, "timestamp": timestamp}
    record.update(measurements)
    return record


record = build_sensor_record(
    "S1", "2024-01-15 08:00:00",
    temperature_c=22.3,
    pressure_kpa=101.2,
    humidity_pct=65.0
)
print(f"\nRecord: {record}")
```

## Solution — Exercise 7: Mutable vs Immutable


```python
# Part A — x = 10, result = 20
def double_value(n):
    n = n * 2
    return n

x = 10
result = double_value(x)
print(x, result)  # 10 20 — x is unchanged; n = n*2 created a new int
```

```python
# Part B — data = [10.0, 20.0, 30.0]
def scale_all_inplace(values, factor):
    for i in range(len(values)):
        values[i] *= factor  # modifies the list object in place

data = [1.0, 2.0, 3.0]
scale_all_inplace(data, 10)
print(data)  # [10.0, 20.0, 30.0] — the original list was modified!
```

```python
# Part C — data = [1.0, 2.0, 3.0], scaled = [10.0, 20.0, 30.0]
def scale_all_new(values, factor):
    return [v * factor for v in values]  # returns a NEW list

data = [1.0, 2.0, 3.0]
scaled = scale_all_new(data, 10)
print(data, scaled)  # data unchanged; scaled is new

# Explanation:
# Part B uses values[i] *= factor which calls __setitem__ on the SAME list object.
# Both 'values' inside the function and 'data' outside refer to the SAME list.
# Part C builds a new list with a comprehension; 'data' and 'values' still
# point to the original, which is never modified.
```

## Solution — Exercise 8: Scope


```python
STEEL_DENSITY = 7850  # kg/m³ — global constant


def compute_mass(volume_m3):
    """
    Compute the mass of a steel component.

    Parameters
    ----------
    volume_m3 : float
        Volume in cubic metres.

    Returns
    -------
    float
        Mass in kilograms.
    """
    density = STEEL_DENSITY  # local variable that shadows... actually reads global
    print(f"  Inside function: local density = {density}")
    return volume_m3 * density


mass = compute_mass(0.01)
print(f"Mass of 0.01 m³ steel: {mass:.1f} kg")

# Trying to access 'density' outside the function
try:
    print(density)  # NameError: 'density' is local to compute_mass
except NameError as e:
    print(f"NameError: {e} — 'density' only exists inside compute_mass()")
```

```python
call_count = 0


def increment_counter():
    """Increment the global call counter."""
    global call_count
    call_count += 1


for _ in range(5):
    increment_counter()

print(f"Call count: {call_count}")  # 5

# Why global state is discouraged:
# Any function can accidentally modify a global variable, making it hard to
# trace where a value changed. Functions with global side effects are harder
# to test and reason about. A better approach: return the updated value and
# let the caller manage state, or use a class to encapsulate mutable state.
```

## Solution — Exercise 9: Lambda Functions


```python
sensor_records = [
    {"sensor_id": "S1", "temperature_c": 22.3, "pressure_kpa": 101.2, "status": "OK"},
    {"sensor_id": "S2", "temperature_c": -999.0, "pressure_kpa": 101.6, "status": "ERROR"},
    {"sensor_id": "S1", "temperature_c": 23.5, "pressure_kpa": 101.0, "status": "OK"},
    {"sensor_id": "S2", "temperature_c": 24.1, "pressure_kpa": 101.7, "status": "OK"},
    {"sensor_id": "S1", "temperature_c": 24.0, "pressure_kpa": 100.9, "status": "OK"},
]

# 1. Lambda to convert Pa → MPa
to_mpa = lambda pa: pa / 1e6
print(f"45000000 Pa = {to_mpa(45e6):.1f} MPa")

# 2. Sort by pressure descending
by_pressure = sorted(sensor_records, key=lambda r: r["pressure_kpa"], reverse=True)
print("\nSorted by pressure (descending):")
for r in by_pressure:
    print(f"  {r['sensor_id']}: {r['pressure_kpa']:.2f} kPa")

# 3. Record with highest temperature
hottest = max(sensor_records, key=lambda r: r["temperature_c"])
print(f"\nHottest record: {hottest['sensor_id']} at {hottest['temperature_c']:.1f} °C")

# 4. Filter to OK records
ok_records = list(filter(lambda r: r["status"] == "OK", sensor_records))
print(f"\nOK records: {len(ok_records)}")

# 5. Map to Fahrenheit
temps_f = list(map(lambda r: r["temperature_c"] * 9/5 + 32, ok_records))
print(f"Fahrenheit temps: {[round(t, 1) for t in temps_f]}")
```

## Solution — Stretch Goal: Engineering Function Library


```python
import math

GRAVITY = 9.81  # m/s²


def hydraulic_diameter(area_m2, wetted_perimeter_m):
    """
    Compute the hydraulic diameter of a conduit cross-section.

    Formula: D_h = 4A / P

    Parameters
    ----------
    area_m2 : float
        Cross-sectional area in m².
    wetted_perimeter_m : float
        Wetted perimeter in metres.

    Returns
    -------
    float
        Hydraulic diameter in metres.

    Raises
    ------
    ValueError
        If area or perimeter are non-positive.
    """
    if area_m2 <= 0:
        raise ValueError(f"area_m2 must be positive, got {area_m2}")
    if wetted_perimeter_m <= 0:
        raise ValueError(f"wetted_perimeter_m must be positive, got {wetted_perimeter_m}")
    return 4 * area_m2 / wetted_perimeter_m


def pipe_velocity(flow_rate_m3s, diameter_m):
    """
    Compute average flow velocity in a circular pipe.

    Formula: v = Q / A = Q / (π × (d/2)²)

    Parameters
    ----------
    flow_rate_m3s : float
        Volumetric flow rate in m³/s.
    diameter_m : float
        Pipe internal diameter in metres.

    Returns
    -------
    float
        Flow velocity in m/s.

    Raises
    ------
    ValueError
        If diameter is non-positive or flow rate is negative.
    """
    if diameter_m <= 0:
        raise ValueError(f"diameter_m must be positive, got {diameter_m}")
    if flow_rate_m3s < 0:
        raise ValueError(f"flow_rate_m3s must be non-negative, got {flow_rate_m3s}")
    area = math.pi * (diameter_m / 2) ** 2
    return flow_rate_m3s / area


def friction_loss(velocity_ms, length_m, diameter_m, friction_factor):
    """
    Compute head loss due to pipe friction using Darcy-Weisbach.

    Formula: h_f = f × (L/D) × (v² / (2g))

    Parameters
    ----------
    velocity_ms : float
        Flow velocity in m/s.
    length_m : float
        Pipe length in metres.
    diameter_m : float
        Pipe internal diameter in metres.
    friction_factor : float
        Darcy friction factor (dimensionless).

    Returns
    -------
    float
        Head loss in metres of fluid.
    """
    if diameter_m <= 0:
        raise ValueError(f"diameter_m must be positive, got {diameter_m}")
    return friction_factor * (length_m / diameter_m) * (velocity_ms ** 2 / (2 * GRAVITY))


def full_pipe_analysis(flow_rate_m3s, diameter_m, length_m, friction_factor,
                       density=998.0, viscosity=1.002e-3):
    """
    Perform a full pipe hydraulic analysis.

    Parameters
    ----------
    flow_rate_m3s : float
        Volumetric flow rate in m³/s.
    diameter_m : float
        Pipe internal diameter in metres.
    length_m : float
        Pipe length in metres.
    friction_factor : float
        Darcy friction factor.
    density : float, optional
        Fluid density in kg/m³ (default: 998 — water at 20°C).
    viscosity : float, optional
        Dynamic viscosity in Pa·s (default: 1.002e-3 — water at 20°C).

    Returns
    -------
    dict
        Summary with velocity, Reynolds number, flow regime, and head loss.
    """
    v = pipe_velocity(flow_rate_m3s, diameter_m)
    area = math.pi * (diameter_m / 2) ** 2
    D_h = hydraulic_diameter(area, math.pi * diameter_m)  # = diameter for circular pipe
    Re = (v * D_h * density) / viscosity

    if Re < 2300:
        regime = "Laminar"
    elif Re <= 4000:
        regime = "Transitional"
    else:
        regime = "Turbulent"

    h_f = friction_loss(v, length_m, diameter_m, friction_factor)

    return {
        "flow_rate_m3s": flow_rate_m3s,
        "velocity_ms": v,
        "reynolds_number": Re,
        "flow_regime": regime,
        "head_loss_m": h_f
    }


# Test
result = full_pipe_analysis(
    flow_rate_m3s=0.01,  # 10 L/s
    diameter_m=0.1,      # 100 mm pipe
    length_m=50.0,
    friction_factor=0.02
)
print("Pipe Analysis Results:")
for key, val in result.items():
    if isinstance(val, float):
        print(f"  {key:<22}: {val:.4f}")
    else:
        print(f"  {key:<22}: {val}")
```
