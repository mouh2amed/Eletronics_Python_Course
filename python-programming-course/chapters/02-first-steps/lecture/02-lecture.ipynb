## 2.1 Arithmetic in Python

Python supports all standard arithmetic operations. The operators and their behaviour are summarised below:

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `3 + 2` | `5` |
| `-` | Subtraction | `10 - 4` | `6` |
| `*` | Multiplication | `4 * 3` | `12` |
| `/` | True (float) division | `7 / 2` | `3.5` |
| `//` | Floor (integer) division | `7 // 2` | `3` |
| `%` | Modulus (remainder) | `7 % 2` | `1` |
| `**` | Exponentiation | `2 ** 8` | `256` |

**Operator precedence** follows standard mathematical convention: `**` first, then `*`, `/`, `//`, `%`, then `+`, `-`. Use parentheses to override.


````python
# Basic arithmetic — Python as an engineering calculator
GRAVITY = 9.81  # m/s², standard gravitational acceleration

mass_kg = 1200.0  # kg, mass of a structural element
weight_n = mass_kg * GRAVITY
print(f"Weight of element: {weight_n:.2f} N")
print(f"Weight of element: {weight_n / 1000:.3f} kN")
````

````python
# Integer vs float division — a critical distinction
total_samples = 7
num_sensors = 2

print(f"True division  : {total_samples / num_sensors}")   # 3.5
print(f"Floor division : {total_samples // num_sensors}")  # 3
print(f"Remainder      : {total_samples % num_sensors}")   # 1
````

````python
# Exponentiation — useful for formulas like V = (4/3) * pi * r^3
import math

radius_m = 0.05  # 5 cm radius sphere
volume_m3 = (4 / 3) * math.pi * radius_m ** 3
print(f"Sphere volume: {volume_m3:.6f} m³")
print(f"Sphere volume: {volume_m3 * 1e6:.4f} cm³")
````

````python
# Operator precedence — parentheses are your friend
# Kinetic energy: E_k = (1/2) * m * v^2
mass_kg = 500.0    # kg
velocity_ms = 15.0  # m/s

# Without parentheses — WRONG (precedence causes error in more complex expressions)
kinetic_wrong = 1 / 2 * mass_kg * velocity_ms ** 2

# With explicit parentheses — CORRECT and clear
kinetic_correct = (1 / 2) * mass_kg * velocity_ms ** 2

print(f"Kinetic energy (both are same here): {kinetic_correct:.1f} J")

# The math module provides constants and functions
print(f"\nUseful math constants:")
print(f"  pi  = {math.pi:.6f}")
print(f"  e   = {math.e:.6f}")
print(f"  sqrt(2) = {math.sqrt(2):.6f}")
````

### Check Your Understanding — 2.1

**Question**: What does `17 % 5` evaluate to, and how might an engineer use the modulus operator?

**Answer**: `17 % 5 = 2`. The modulus returns the remainder after integer division (17 = 3 × 5 + 2). Engineers use it to check periodicity (e.g., every 10th reading), wrap indices, or check divisibility.


````python
# Engineering use of modulus: flag every 4th reading for calibration check
for reading_number in range(1, 13):
    if reading_number % 4 == 0:
        print(f"Reading {reading_number:2d}: CALIBRATION CHECK")
    else:
        print(f"Reading {reading_number:2d}: normal")
````

## 2.2 Output and Printing

The `print()` function sends output to the screen (standard output). It accepts any number of arguments and has two important keyword parameters:

- `sep` — separator between multiple arguments (default: a single space `' '`).
- `end` — string appended at the end (default: newline `'\n'`).


````python
# print() with multiple arguments
sensor_id = "S1"
temperature = 22.8
unit = "°C"

print(sensor_id, temperature, unit)                      # default sep = ' '
print(sensor_id, temperature, unit, sep=", ")            # custom separator
print(sensor_id, temperature, unit, sep=" | ")           # pipe separator
````

````python
# end parameter — suppress the automatic newline
print("Loading", end="")
print(".", end="")
print(".", end="")
print(".")  # final print restores newline

# More useful: print items on one line from a loop
print("Sensors active: ", end="")
for s in ["S1", "S2", "S3"]:
    print(s, end=" ")
print()  # add a final newline
````

````python
# f-string formatting — the preferred approach in Python 3.6+
beam_length_m = 6.35
load_kn = 42.753
deflection_mm = 3.2

print("--- Beam Calculation Summary ---")
print(f"Span       : {beam_length_m:.2f} m")
print(f"Load       : {load_kn:.1f} kN")
print(f"Deflection : {deflection_mm:.3f} mm")

# Alignment using format specifiers
print("\n--- Aligned table ---")
print(f"{'Parameter':<15} {'Value':>10} {'Unit':<8}")
print("-" * 35)
print(f"{'Span':<15} {beam_length_m:>10.2f} {'m':<8}")
print(f"{'Load':<15} {load_kn:>10.1f} {'kN':<8}")
print(f"{'Deflection':<15} {deflection_mm:>10.3f} {'mm':<8}")
````

**f-string format specifiers** follow this pattern inside `{}`:

```
{value:[fill][align][width][.precision][type]}
```

| Specifier | Meaning | Example |
|---|---|---|
| `:.2f` | 2 decimal places, fixed-point | `{3.14159:.2f}` → `3.14` |
| `:>10` | right-align in width 10 | `{42:>10}` → `        42` |
| `:<10` | left-align in width 10 | `{'hi':<10}` → `hi        ` |
| `:^10` | centre in width 10 | `{'hi':^10}` → `    hi    ` |
| `:e` | scientific notation | `{12345.6:e}` → `1.234560e+04` |
| `:,.2f` | thousands separator | `{1234567.8:,.2f}` → `1,234,567.80` |

## 2.3 Variables and Types

A **variable** is a named reference to a value stored in memory. In Python, you assign a value with `=`:

```python
temperature_celsius = 22.5
```

Python is **dynamically typed**: you do not declare the type; Python infers it from the value assigned. The four primitive types used constantly:

| Type | Description | Example |
|---|---|---|
| `int` | Integer (whole number) | `42`, `-7`, `0` |
| `float` | Floating-point number | `3.14`, `-0.001`, `9.81` |
| `str` | String (text) | `'S1'`, `"pressure"` |
| `bool` | Boolean | `True`, `False` |


````python
# Variable assignment and type inspection
beam_length_m = 5.0       # float
num_supports = 2          # int
material = "steel"        # str
is_prestressed = False    # bool

print(f"{beam_length_m} is of type {type(beam_length_m).__name__}")
print(f"{num_supports} is of type {type(num_supports).__name__}")
print(f"{material!r} is of type {type(material).__name__}")
print(f"{is_prestressed} is of type {type(is_prestressed).__name__}")
````

````python
# Type conversion (casting)
raw_value = "22.5"     # string received from sensor (as text)

temperature = float(raw_value)   # convert string → float
print(f"Temperature: {temperature:.1f} °C (type: {type(temperature).__name__})")

# int() truncates towards zero — it does not round
print(f"int(3.9) = {int(3.9)}")   # 3, not 4
print(f"int(-3.9) = {int(-3.9)}") # -3, not -4

# str() converts numbers to strings
sample_id = 7
label = "Sample_" + str(sample_id)  # string concatenation requires same type
print(f"Label: {label}")
````

````python
# None — the absence of a value
# Used to initialise a variable before its value is known
last_reading = None
print(f"last_reading is None: {last_reading is None}")

# After the sensor provides a value:
last_reading = 23.1
print(f"last_reading is None: {last_reading is None}")
print(f"last_reading = {last_reading}")
````

### Check Your Understanding — 2.3

**Question**: What does `type(True)` return? Is `True` equal to `1` in Python?

**Answer**: `type(True)` returns `<class 'bool'>`. Python's `bool` is a subclass of `int`. `True == 1` evaluates to `True`, and `False == 0` evaluates to `True`. This means `True + True == 2`. While this works, relying on it is considered poor style — use explicit integer values when you mean integers.

## 2.4 Strings

Strings are sequences of characters. They are **immutable** — once created, you cannot change individual characters; you create a new string instead.

### String Literals

```python
s1 = 'single quotes'
s2 = "double quotes"          # identical to single quotes
s3 = "It's an f-string"       # single quote inside double-quoted string
s4 = 'He said "hello"'        # double quote inside single-quoted string
s5 = """triple-quoted
spans multiple lines"""        # triple quotes allow embedded newlines
```


````python
# String indexing — zero-based
sensor_label = "SENSOR_S1"
#                0123456789

print(f"First character  : {sensor_label[0]}")
print(f"Last character   : {sensor_label[-1]}")
print(f"Sixth character  : {sensor_label[5]}")
````

````python
# String slicing — s[start:stop:step] — stop is EXCLUSIVE
log_line = "2024-01-15 08:00:00 | INFO | System started"

date_part = log_line[0:10]     # characters 0 through 9
time_part = log_line[11:19]    # characters 11 through 18
level = log_line[22:26].strip()  # log level

print(f"Date  : {date_part}")
print(f"Time  : {time_part}")
print(f"Level : {level}")
````

````python
# String methods — built-in operations on strings
raw_input = "  22.5 OK  "  # typical messy sensor string

print(f"Original    : {raw_input!r}")
print(f"stripped    : {raw_input.strip()!r}")         # remove leading/trailing whitespace
print(f"upper       : {raw_input.strip().upper()!r}") # uppercase
print(f"lower       : {raw_input.strip().lower()!r}") # lowercase

# split() — divide a string by a delimiter
csv_line = "S1,22.3,101.2,OK"
fields = csv_line.split(",")
print(f"\nCSV split: {fields}")
print(f"Sensor ID  : {fields[0]}")
print(f"Temperature: {float(fields[1])} °C")
````

````python
# find() and replace()
log_entry = "[2024-01-15 08:16:00] ERROR  Failed to read S2"

pos = log_entry.find("ERROR")
print(f"'ERROR' found at index: {pos}")   # -1 if not found

fixed = log_entry.replace("ERROR", "WARNING")
print(f"Modified: {fixed}")

# len() — number of characters
print(f"Length of log entry: {len(log_entry)} characters")
````

## 2.5 Loops and Conditionals

### Conditional Statements — if / elif / else

The `if` statement evaluates a boolean condition. If `True`, the indented block runs. `elif` (else if) provides additional conditions. `else` catches everything not caught above.


````python
# Structural design check: classify a beam's load ratio
applied_load_kn = 85.0
allowable_load_kn = 100.0

load_ratio = applied_load_kn / allowable_load_kn

if load_ratio < 0.5:
    status = "Under-utilised"
elif load_ratio < 0.85:
    status = "Acceptable"
elif load_ratio <= 1.0:
    status = "Near capacity — review"
else:
    status = "OVERSTRESSED — redesign required"

print(f"Applied / Allowable = {load_ratio:.2%} → {status}")
````

````python
# Boolean operators: and, or, not
temperature_c = 24.0
pressure_kpa = 101.5

TEMP_MIN = 0.0
TEMP_MAX = 40.0
PRESSURE_MIN = 95.0
PRESSURE_MAX = 110.0

temp_ok = TEMP_MIN <= temperature_c <= TEMP_MAX          # chained comparison
pressure_ok = PRESSURE_MIN <= pressure_kpa <= PRESSURE_MAX

if temp_ok and pressure_ok:
    print("All readings within normal range.")
elif not temp_ok:
    print(f"Temperature out of range: {temperature_c} °C")
elif not pressure_ok:
    print(f"Pressure out of range: {pressure_kpa} kPa")
````

````python
# for loop — iterate over a range of numbers
# Example: compute the deflection at multiple load increments
# Simple formula: delta = F * L^3 / (48 * E * I)

E = 200e9    # Pa, Young's modulus for steel
I = 8.33e-5  # m^4, second moment of area
L = 5.0      # m, beam span

print(f"{'Load (kN)':<12} {'Deflection (mm)':<18}")
print("-" * 30)

for load_kn in range(10, 110, 10):  # 10, 20, ..., 100 kN
    F = load_kn * 1000  # convert kN → N
    deflection_m = F * L**3 / (48 * E * I)
    deflection_mm = deflection_m * 1000
    print(f"{load_kn:<12} {deflection_mm:<18.4f}")
````

````python
# for loop — iterate over a list
sensor_ids = ["S1", "S2", "S3", "S4"]
readings = [22.3, -999.0, 23.1, 22.8]  # -999.0 is a fault marker

FAULT_VALUE = -999.0

for i, sensor in enumerate(sensor_ids):
    reading = readings[i]
    if reading == FAULT_VALUE:
        print(f"{sensor}: FAULT — reading discarded")
    else:
        print(f"{sensor}: {reading:.1f} °C")
````

````python
# while loop — repeat while a condition holds
# Example: iterative calculation until convergence

# Newton's method: approximate sqrt(N) without math.sqrt
N = 144.0
guess = N / 2.0      # initial guess
tolerance = 1e-8     # convergence threshold
iteration = 0

while abs(guess ** 2 - N) > tolerance:
    guess = (guess + N / guess) / 2.0
    iteration += 1

print(f"sqrt({N}) ≈ {guess:.10f}  (converged in {iteration} iterations)")
print(f"math.sqrt gives: {math.sqrt(N):.10f}")
````

````python
# break and continue
temperatures = [21.5, 22.0, 23.1, -999.0, 24.5, 22.8]

print("Processing readings (continue skips faults):")
for t in temperatures:
    if t == -999.0:
        print(f"  Fault detected — skipping.")
        continue          # jump to next iteration without processing
    print(f"  Temperature: {t:.1f} °C")

print("\nProcessing readings (break stops at first fault):")
for t in temperatures:
    if t == -999.0:
        print(f"  Fault detected — stopping processing.")
        break             # exit the loop entirely
    print(f"  Temperature: {t:.1f} °C")
````

## 2.6 Reading Keyboard Input

`input()` pauses execution and waits for the user to type something and press Enter. It **always returns a string** — you must convert the result if you need a number.


````python
# In a .py script, this reads from the keyboard:
#   name = input("Enter your name: ")
#   print(f"Hello, {name}!")

# In Jupyter, input() works in some kernels but can be unreliable.
# We demonstrate the pattern here with a fixed value:

# --- Pattern 1: Read a string ---
name = "Alice"   # simulates: name = input("Enter your name: ")
print(f"Hello, {name}!")

# --- Pattern 2: Read a number ---
# CRITICAL: input() always returns a string; you MUST convert to int or float
raw = "25"       # simulates: raw = input("Enter beam span in metres: ")
span_m = float(raw)   # convert string to float
print(f"Beam span: {span_m:.1f} m")
````

````python
# Full interactive unit converter (run as a .py script for live input)
# This cell shows the complete program structure:

def celsius_to_fahrenheit(temp_c):
    """Convert Celsius to Fahrenheit."""
    return temp_c * 9 / 5 + 32

# In a real script:
#   raw = input("Enter temperature in Celsius: ")
#   temp_c = float(raw)

# Demonstration with a fixed value:
temp_c = 100.0
temp_f = celsius_to_fahrenheit(temp_c)
print(f"{temp_c:.1f} °C = {temp_f:.1f} °F")
````
